# Benchmark with a constant tar formula

Balance-and-correlation models represent the tar with a constant formula, here C6H6.2O0.2 (Thunman et al., 2001), and weight the elemental balances as residuals (Section 4.2). Variant A takes the char composition from the correlations of Neves et al. (2011); variant B uses the carbonisation closure of the framework. The check with exact elimination asks whether any admissible state with the constant formula exists.

```python
import os
import shutil
import sys
import tempfile
from pathlib import Path
ARTICLE = Path.cwd().parent / "results"
OUTPUT = Path.cwd() / "output" / "06_balance_and_correlation"
OUTPUT.mkdir(parents=True, exist_ok=True)
if (ARTICLE / "best_vectors.npz").exists():
    handle, temporary = tempfile.mkstemp(dir=OUTPUT, suffix=".npz")
    os.close(handle)
    shutil.copyfile(ARTICLE / "best_vectors.npz", temporary)
    os.replace(temporary, OUTPUT / "best_vectors.npz")
os.environ.setdefault("CCPYRO_RESULTS", str(OUTPUT))
os.environ.setdefault("CCPYRO_FIGDATA", str(OUTPUT / "figdata"))
try:
    import ccpyro  # installed with `pip install -e .`
except ImportError:  # running from a clone without installation
    sys.path.insert(0, str(Path.cwd().parent / "src"))
    import ccpyro
import numpy as np
import pandas as pd
np.set_printoptions(precision=6, suppress=True)
print("ccpyro", ccpyro.__version__)
```

```python
from ccpyro import correlations as cor
table, check = cor.run(OUTPUT, cases=[("rice husk", "pelletized")], starts=8)
check
```
